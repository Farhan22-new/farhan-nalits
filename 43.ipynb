```python
#97
def pascal_row(n):
    row = [1]

    for k in range(1, n + 1):
        value = row[k - 1] * (n - k + 1) // k
        row.append(value)

    return row


n = int(input("Enter row number: "))

print("Pascal row:", pascal_row(n))
```

```python
#99
def is_pangram(sentence):

    alphabet = "abcdefghijklmnopqrstuvwxyz"

    sentence = sentence.lower()

    for letter in alphabet:
        if letter not in sentence:
            return False

    return True


sentence = input("Enter a sentence: ")

print("Is Pangram?:", is_pangram(sentence))
```

```python
#100
def standard_deviation(numbers):

    mean = sum(numbers) / len(numbers)

    total = 0

    for number in numbers:
        total = total + (number - mean) ** 2

    variance = total / len(numbers)

    standard_deviation = variance ** 0.5

    return standard_deviation


values = input("Enter numbers separated by spaces: ").split()

numbers = []

for value in values:
    numbers.append(float(value))

result = standard_deviation(numbers)

print("Standard Deviation:", result)
```

```python
#101
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []

print("Enter first matrix:")

for i in range(rows):
    row = input("Enter row values separated by spaces: ").split()

    number_row = []

    for value in row:
        number_row.append(int(value))

    matrix1.append(number_row)


print("Enter second matrix:")

for i in range(rows):
    row = input("Enter row values separated by spaces: ").split()

    number_row = []

    for value in row:
        number_row.append(int(value))

    matrix2.append(number_row)


result = []

for i in range(rows):

    result_row = []

    for j in range(columns):
        result_row.append(matrix1[i][j] + matrix2[i][j])

    result.append(result_row)


print("Result:", result)
```

```python
#102
rows_a = int(input("Enter rows of Matrix A: "))
cols_a = int(input("Enter columns of Matrix A: "))

rows_b = int(input("Enter rows of Matrix B: "))
cols_b = int(input("Enter columns of Matrix B: "))

if cols_a != rows_b:
    print("Matrices cannot be multiplied")

else:
    A = []
    B = []

    print("Enter Matrix A:")

    for i in range(rows_a):
        row = input("Enter row values: ").split()

        number_row = []

        for number in row:
            number_row.append(int(number))

        A.append(number_row)


    print("Enter Matrix B:")

    for i in range(rows_b):
        row = input("Enter row values: ").split()

        number_row = []

        for number in row:
            number_row.append(int(number))

        B.append(number_row)


    result = []

    for i in range(rows_a):
        result_row = []

        for j in range(cols_b):
            total = 0

            for k in range(cols_a):
                total = total + A[i][k] * B[k][j]

            result_row.append(total)

        result.append(result_row)


    print("Result:", result)
```

```python
#104
numbers = input("Enter numbers separated by spaces: ").split()

number_list = []

for number in numbers:
    number_list.append(int(number))

n = len(number_list)

for i in range(n):

    for j in range(0, n - i - 1):

        if number_list[j] > number_list[j + 1]:

            number_list[j], number_list[j + 1] = number_list[j + 1], number_list[j]

print("Sorted list:", number_list)
```

```python
#105
numbers = input("Enter sorted numbers separated by spaces: ").split()

number_list = []

for number in numbers:
    number_list.append(int(number))

target = int(input("Enter target number: "))

low = 0
high = len(number_list) - 1
found = -1

while low <= high:

    middle = (low + high) // 2

    if number_list[middle] == target:
        found = middle
        break

    elif number_list[middle] < target:
        low = middle + 1

    else:
        high = middle - 1


if found != -1:
    print("Target found at index:", found)

else:
    print("Target not found")
```
