```python
target =int(input("Enter a value:"))

val1 = int(input("Enter a number:"))
val2 = int(input("Enter a number again:"))

diff1=abs(val1-target) 
diff2=abs(val2-target)

if diff1 < diff2:
    print(val1, " val1 is  closer")
elif diff2<diff1:
    print(val2, "val2 is closer")
else:
    print("equally closer")
```
