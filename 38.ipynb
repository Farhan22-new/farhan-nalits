```python
num = int(input("Enter a number: "))

is_prime = True

if num <= 1:
    is_prime = False
else:
    for number in range(2, num):
        if num % number == 0:
            is_prime = False
            break

if is_prime == True:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```
